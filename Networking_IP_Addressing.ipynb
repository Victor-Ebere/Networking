# **WHAT I LEARNT**

An IP address is the logical address of a network interface that allows other computers, devices to commuicate with it over an IP network. And an IP  stands for internet protocol. There are two types of IP, IPv4 and IPv6. My computer is uses an IPv4 address.

wlan0 represents my wifi network interface.

Default gateway is a router on a local network that handles data when a device needs to communicate with an outside network, such as the internet.

127.0.0.1 is a loopback. It is the address of the network interface with which my computer uses to communicate to itself(localhost).

ip addr displays network interfaces and the IP addresses assigned to them.

ip route displays the routing table, including the default gateway and the networks directly reachable through an interface.

ping can be used to test IP connectivity between the computer and another host.

wlan0 is my wireless network interface, while eth0 represents my Ethernet interface.

127.0.0.1 is the IPv4 loopback address used for communication within the local computer.

10.75.240.26/24 is my current private IPv4 address on wlan0.

10.75.240.138 is my default gateway.
