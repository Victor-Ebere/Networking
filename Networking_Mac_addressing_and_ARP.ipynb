1. MAC Address

A MAC address is the hardware/link-layer address associated with a network interface.

Unlike an IP address, which is a logical address, the MAC address is used for communication on the local network.

A network interface and a MAC address are not the same thing.

The interface is the connection point, while the MAC address is the link-layer/hardware address associated with that interface.

# **IP Address vs MAC Address**

An IP address is the logical address of a network interface that allows devices to communicate over an IP network.

A MAC address is the hardware/link-layer address used for communication on the local network.

The IP tells us where the communication is logically going, while the MAC is used to deliver the frame to the appropriate device/interface on the current link.


# **ARP**

ARP means Address Resolution Protocol.

From the name itself, I understood that ARP solves the problem of a computer knowing the IP address it wants to communicate with but not knowing the specific hardware/MAC address needed to deliver the frame on the local network.

My analogy for ARP is a dispatch rider.

The rider knows the location/address of where the package is going, but doesn't necessarily know who exactly to deliver it to.

So ARP basically solves:"I know the IP address, but what MAC address do I use to deliver this on my local network?"

i.e My computer knows my default gateway IP is 10.75.240.138

ARP resolves the IP to 2a:bf:6e:81:50:4d

This tells me that my computer knows the relationship between

MAC: 2a:bf:6e:81:50:4d

The REACHABLE state means my system currently considers that neighbor reachable.

ip neigh doesn't actually perform the ARP request.

Instead, it allows me to see the neighbor information my system has learned/stored.

ARP is the process.

ip neigh is one way of viewing the result.


# PACKET VS FRAME

A packet is what carries the IP communication from the source toward the final destination.

A frame carries the packet across an individual network link/hop.

The MAC addresses can change from hop to hop, while the IP packet continues toward its destination.
