# **ROUTING, PACKETS & FRAMES**

1. **ROUTING TABLE**

The routing table tells my laptop where to send packets based on their destination IP.



```
ip route
```
image.png

This shows that my default gateway is 10.75.240.138, and my laptop uses wlan0 to reach it.

If the destination isn't on my local network, my computer uses the default route.



2. **NEXT HOP**

```
ip route get 8.8.8.8
```
image.png

This means 8.8.8.8 is the destination, while 10.75.240.26 is the next hop/default gateway.


3. **ROUTING + ARP**

Routing tells my computer where to send the packet next, while ARP finds the MAC address needed to deliver the frame on the local network.

I deleted my gateway's MAC from the neighbour cache with this command

```
sudo ip neigh del 10.75.240.138 dev wlan0
```
image.png


and my computer ARPed for it again, we can see this using

```
ip neigh
```

image.png

I also used tcpdump to watch the ARP request and reply.

image.png

4. **PACKET AND FRAME**

A packet uses IP addresses to communicate from the source toward the final destination.

A frame carries the packet from one hop to the next using MAC addresses.

The IP destination can remain 8.8.8.8, while the MAC destination is my gateway's MAC for the first hop.

5. **WHAT I LEARNT/OBSERVED**

Using
```
sudo tcpdump -eni wlan0 'icmp and host 8.8.8.8'
```
on one terminal and running
```
ping -c 4 8.8.8.8
```
on the second terminal.


image.png

I observed packets nSRC 10.75.240.26 > DST 8.8.8.8

while the Frame was addressed to my gateway's MAC.

Which solidified my understanding that IP tells the packet where it ultimately needs to go, while MAC tells the frame where to go on the current hop.
